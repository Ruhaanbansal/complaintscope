## 5. Model Explainability & Global Feature Attributions

Logistic regression coefficients reveal the specific vocabulary terms most predictive of affirmative consumer relief versus no relief.

```python
feat_path = Path('../results/metrics/relief_global_features.json') if Path('../results/metrics/relief_global_features.json').exists() else Path('results/metrics/relief_global_features.json')
if feat_path.exists():
    with open(feat_path) as f:
        feats = json.load(f)
    df_pos = pd.DataFrame(feats.get('top_positive_features', []))
    df_neg = pd.DataFrame(feats.get('top_negative_features', []))
    print('Top 10 Positive Relief Indicators (Higher Relief Likelihood):')
    print(df_pos.head(10).to_string(index=False))
    print('\nTop 10 Negative Relief Indicators (Lower Relief Likelihood):')
    print(df_neg.head(10).to_string(index=False))
else:
    print('Global feature metrics not found yet.')
```
