## 1. Systemic Risk Index Composite Formulation

The SRI synthesizes five normalized regulatory dimensions:
$$\text{SRI} = w_1 \cdot \text{Burst} + w_2 \cdot \text{Growth} + w_3 \cdot \text{Breadth} + w_4 \cdot \text{ReliefRate} + w_5 \cdot \text{Volume}$$

Default baseline weights: $\{0.25, 0.25, 0.20, 0.15, 0.15\}$.

```python
sri_path = Path('../results/tables/sri_rankings.csv') if Path('../results/tables/sri_rankings.csv').exists() else Path('results/tables/sri_rankings.csv')
if sri_path.exists():
    df_sri = pd.read_csv(sri_path)
    print('Top Institutions Ranked by Systemic Risk Index (SRI):')
    print(df_sri.head(20).to_string(index=False))
else:
    print('SRI rankings table not found yet.')
```
