## 2. Community Detection & Configuration Null Model Hypothesis Test

To rigorously confirm whether observed institutional communities reflect true structural clustering rather than random chance, we execute a degree-preserving configuration null model (100 iterations).

```python
null_path = Path('../results/metrics/network_null_model.json') if Path('../results/metrics/network_null_model.json').exists() else Path('results/metrics/network_null_model.json')
if null_path.exists():
    with open(null_path) as f:
        null_data = json.load(f)
    print('Configuration Null Model Results:')
    print(f"  Observed Modularity:    {null_data.get('observed_modularity', 'N/A')}")
    print(f"  Null Mean Modularity:   {null_data.get('null_mean_modularity', 'N/A')}")
    print(f"  Empirical p-value:      {null_data.get('null_pvalue', 'N/A')}")
    print(f"  Iterations:             {null_data.get('n_iterations', 'N/A')}")
else:
    print('Null model results file not found yet.')
```
