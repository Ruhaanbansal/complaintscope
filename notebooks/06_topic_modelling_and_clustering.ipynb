## 3. Discovered Topic Catalog & Top Salient Words

Domain-labelled topics extracted from the model's top word distributions.

```python
labels_path = Path('../results/metrics/topic_labels.json') if Path('../results/metrics/topic_labels.json').exists() else Path('results/metrics/topic_labels.json')
if labels_path.exists():
    with open(labels_path, 'r', encoding='utf-8') as f:
        topics_data = json.load(f)
    print('Discovered Topic Taxonomy:')
    for model_name, tdict in topics_data.items():
        print(f'\nModel: {model_name.upper()}')
        for tid, tinfo in tdict.items():
            words_str = ', '.join(tinfo.get('top_words', [])[:7])
            print(f"  Topic {tid} [{tinfo.get('label')}]: {words_str}")
else:
    print('Topic labels file not found yet.')
```
